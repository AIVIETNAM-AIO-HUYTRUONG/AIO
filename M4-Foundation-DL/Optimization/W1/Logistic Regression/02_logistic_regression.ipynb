# **1. Tự suy ra đạo hàm của Logistic Regression**

- Chúng ta đã biết các công thức quan trọng. Nhưng nếu chỉ ghi nhớ hai công thức này, chúng ta sẽ khó hiểu tại sao chúng đúng và khó áp dụng tư duy tương tự cho những mô hình khác.

- Phần này sẽ giúp ta **tự suy ra hai công thức trên theo từng bước**, theo đúng chuỗi toán học **Logistic Regression**: Mô hình → Sigmoid → Binary Cross-Entropy → Chain Rule.

## **1.1. Bước 1 - Tính đạo hàm của Binary Cross-Entropy**

- $\frac{\partial L}{\partial\hat y}$ là đạo hàm của **Loss** theo xác suất dự đoán. **Loss của một mẫu** là:
$$ \boxed{L=-y\log(\hat y)-(1-y)\log(1-\hat y)} $$

- Để dễ theo dõi, ta tách công thức thành hai phần:
$$
L=
\underbrace{-y\log\hat y}_{\text{Phần 1}}
\underbrace{-(1-y)\log(1-\hat y)}_{\text{Phần 2}}
$$



- Ta có:
$ \frac{\partial}{\partial\hat y}(-y\log\hat y) $

- Vì $y$ là hằng số đối với phép lấy đạo hàm theo $\hat y$, ta đưa nó ra ngoài:
$ =-y\frac{\partial}{\partial\hat y}\log\hat y $

- Dùng quy tắc:
$ \frac{d}{du}\log u=\frac{1}{u} $

- Ta được:
$ \boxed{-\frac{y}{\hat y}} $

- Xét:
$$
\frac{\partial}{\partial\hat y}
[-(1-y)\log(1-\hat y)]
$$

- Tiếp tục xem $1-y$ là hằng số:
$$
=-(1-y)\frac{\partial}{\partial\hat y}
\log(1-\hat y)
$$

- Đây là chỗ cần cẩn thận. Theo quy tắc đạo hàm hàm hợp:
$$
\frac{d}{du}\log u=\frac{1}{u}
$$

- Nhưng biểu thức bên trong log là $1-\hat y$, có đạo hàm bằng $-1$. Vì thế:
$$
\frac{\partial}{\partial\hat y}\log(1-\hat y)
=\frac{-1}{1-\hat y}
$$

- Thay lại:
$$
=-(1-y)\left(\frac{-1}{1-\hat y}\right)
$$

- Hai dấu âm triệt tiêu:
$$
\boxed{\frac{1-y}{1-\hat y}}
$$

- Cộng hai kết quả:
$$
\boxed{
\frac{\partial L}{\partial\hat y}
=-\frac{y}{\hat y}
+\frac{1-y}{1-\hat y}
}
$$

## **1.2. Bước 2 - Tính đạo hàm của Sigmoid**

- Nhắc lại định nghĩa:
$$
\hat y=\sigma(z)=\frac{1}{1+e^{-z}}
$$

- Xét:
$$
\frac{\partial}{\partial z}(1+e^{-z})^{-1}
$$

- Để dễ nhìn, đặt:
$$
u=1+e^{-z}
$$

- Khi đó biểu thức ban đầu trở thành:
$$
\frac{d}{dz}u^{-1}
$$

- Theo quy tắc đạo hàm lũy thừa:
$$
\frac{d}{dz}u^{-1}
=-1\cdot u^{-2}\cdot\frac{du}{dz}
$$

- Bây giờ, thay $u=1+e^{-z}$ trở lại (*):
$$
\boxed{
\frac{\partial\hat y}{\partial z}
=-(1+e^{-z})^{-2}
\frac{\partial}{\partial z}(1+e^{-z})
}
$$

- Ta cần tính:
$$
\frac{\partial}{\partial z}(1+e^{-z})
$$

- Đạo hàm của tổng bằng tổng các đạo hàm:
$$
=\frac{\partial}{\partial z}(1)
+\frac{\partial}{\partial z}(e^{-z})
$$

- Đạo hàm của hằng số $1$ bằng $0$. Đối với $e^{-z}$, ta dùng công thức:
$$
\frac{d}{dz}e^{u(z)}
=e^{u(z)}\cdot u'(z)
$$

- Ở đây $u(z)=-z$, nên $u'(z)=-1$. Do đó:
$$
\frac{\partial}{\partial z}(e^{-z})
=e^{-z}\cdot(-1)=-e^{-z}
$$

- Vậy:
$$
\frac{\partial}{\partial z}(1+e^{-z})
=-e^{-z}
$$

- Thay kết quả vào (*) ta sẽ có:
$$
\frac{\partial\hat y}{\partial z}
=-(1+e^{-z})^{-2}(-e^{-z})
$$

- Hai dấu âm nhân với nhau thành dấu dương:
$$
\boxed{
\frac{\partial\hat y}{\partial z}
=\frac{e^{-z}}{(1+e^{-z})^2}
}
$$

## **1.3. Bước 3 - Đạo hàm của phương trình tuyến tính**

- Ta có: $ z=wx+b $

- Lấy đạo hàm theo $w$, với $x$ và $b$ được xem là hằng số:
$ \boxed{\frac{\partial z}{\partial w}=x} $

- Tương tự, lấy đạo hàm theo $b$:
$ \boxed{\frac{\partial z}{\partial b}=1} $

------------------------------------------
- Vậy chúng ta đã có đủ ba thành phần:

| Thành phần | Công thức đạo hàm |
| :--- | :--- |
| **BCE theo dự đoán** | $\frac{\partial L}{\partial\hat{y}} = -\frac{y}{\hat{y}} + \frac{1-y}{1-\hat{y}}$ |
| **Sigmoid theo đầu vào** | $\frac{\partial\hat{y}}{\partial z} = \hat{y}(1-\hat{y})$ |
| **Linear function theo trọng số** | $\frac{\partial z}{\partial w} = x$ |

## **1.4. Bước 4 - Dùng Chain Rule để suy ra gradient**

- Theo Chain Rule:
$$
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial\hat y}
\frac{\partial\hat y}{\partial z}
\frac{\partial z}{\partial w}
$$

- Thay ba thành phần đã tính ở trên:
$$
\frac{\partial L}{\partial w}
=
\left(
-\frac{y}{\hat y}
+\frac{1-y}{1-\hat y}
\right)
\hat y(1-\hat y)x
$$

- Ta sẽ rút gọn biểu thức trong ngoặc trước, rồi sau đó quy đồng mẫu số: $$
-\frac{y}{\hat y}
+\frac{1-y}{1-\hat y}
=
\frac{-y(1-\hat y)+(1-y)\hat y}
{\hat y(1-\hat y)}
$$

- Khai triểu tử số: $-y+y\hat y+\hat y-y\hat y$, hai thành phần $y\hat y$ và $-y\hat y$ triệt tiêu, còn lại:
$$
\hat y-y
$$

- Do đó $$
\frac{\partial L}{\partial\hat y}
=
\frac{\hat y-y}{\hat y(1-\hat y)}
$$

- Hai thành phần $\hat y(1-\hat y)$ triệt tiêu. Kết quả cuối cùng là:
$$
\boxed{\frac{\partial L}{\partial w}=x(\hat y-y)}
$$

## **1.5. Bước 5 - Đạo hàm theo bias**
- Ta cũng dùng Chain Rule:
$$
\frac{\partial L}{\partial b}
=
\frac{\partial L}{\partial\hat y}
\frac{\partial\hat y}{\partial z}
\frac{\partial z}{\partial b}
$$

- Thay các thành phần vào:
$$
=
\frac{\hat y-y}{\hat y(1-\hat y)}
\cdot\hat y(1-\hat y)\cdot1
$$

- Rút gọn:
$$
\boxed{\frac{\partial L}{\partial b}=\hat y-y}
$$
