# Convert Python list to pandas Series

- Một pandas Series lưu một dãy giá trị với một index

```python
cities = [
    "HCM",
    "HN",
    "HN",
    "ĐN",
    "HCM",
    "Đa Nang",
]

cities_series = pd.Series(cities)
cities_series
```

# Inspect a Series
- Kết hợp values, index và name để tạo một Series đầy đủ

```python
order_ids = ["O101", "O102", "O103", "O104", "O105","O106"]
cities_series = pd.Series(cities, name="Cities", index=order_ids)
cities_series
```

# Select by Label

- Một label trả về một giá trị. `loc["a":"b"]` trả về một slice theo label.

```python
value = cities_series.loc["O101"]

print("Value:", value)
print("Type:", type(value))
```

```python
value_slice = cities_series.loc["O101":"O103"]

print("Value:\n", value_slice)
print("Type:", type(value_slice))
```

# Select from a Series
- Một label trả về một giá trị. Một danh sách label trả về một Series.

```python
value = cities_series.loc["O104"]

print("Value:", value)
print("Type:", type(value))
```

```python
value = cities_series.loc[["O104", "O106"]]

print("Value:\n", value)
print("Type:", type(value))
```
