#  Remove Duplicate Rows
- `duplicated()` đánh dấu các bản ghi lặp lại sau bản ghi đầu tiên.
- `drop_duplicates()` giữ lại bản ghi đầu tiên.

```python
duplicate_mask =  exam_results.duplicated()
exam_results[duplicate_mask]
```

```python
df_clean = exam_results.drop_duplicates().copy()
df_clean
```

```python
df_clean.duplicated().sum()
```

# Standardize Text Labels

- Áp dụng một quy tắc chuẩn hóa cho từng cột để tránh một lớp hoặc trạng thái bị tách thành nhiều nhóm khác nhau.

```python
df_clean['Class'].unique()
```

```python
df_clean['Class'] = df_clean['Class'].str.replace(" ", "", regex=False).str.upper()
df_clean['Class'].unique()
```

```python
df_clean["Status"].unique()
```

```python
df_clean["Status"] = df_clean["Status"].str.title()
df_clean["Status"].unique()
```

# Convert Data Type

- Chuyển Scorce sang dạng số và ExamDate sang dạng ngày tháng.

```python
df_clean["ExamDate"] = pd.to_datetime(
    df_clean["ExamDate"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)
df_clean["ExamDate"].unique()
```

```python
df_clean["Score"] = pd.to_numeric(
    df_clean["Score"].astype("string").str.replace("points", "", regex=False),
    errors="coerce"
)

df_clean["Score"].unique()
```

# Handle the Missing Score

- Giá trị Score bị thiếu của học sinh vắng thi được giữa nguyên (không xóa vì phù hớp với Status).

```python
absent_mask = df_clean["Status"].eq("Absent")
df_clean.loc[absent_mask, "Score"]
```

```python
df_clean.loc[absent_mask, "Score"].isna().all()
```

# Prepare Data for Analysis
- Lọc chỉ giữ các bài thi đã hoàn thành (Completed), sau đó chọn các cột cần thiết.

```python
completed_column = df_clean['Status'].eq("Completed")

df_analysis = df_clean.loc[completed_column, ["StudentID","Class", "Score", "ExamDate"]].copy()
df_analysis
```

# Create the Passed Column
- Quy tắc: Passed = Score >= 5.

```python
df_analysis["Passed"] = df_analysis['Score'].ge(5)
df_analysis
```

# Summarize Results by Class
- Nhóm học sinh theo lớp, tính điểm trung bình và số học sinh đạt bằng `agg()`.

```python
class_summary = df_analysis.groupby("Class").agg(
    AverageScore=("Score", "mean"),
    NumberOfPassed=("Passed", "sum")
).round(2)

class_summary
```
