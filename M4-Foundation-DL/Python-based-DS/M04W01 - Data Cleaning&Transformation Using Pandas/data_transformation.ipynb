# From Cleaned Data to an Analysis Question
- Câu hỏi phân tích: Tổng số lượng sản phẩm được đặt ở mỗi thành phố là bao nhiêu?


```python
df_clean = pd.read_csv("/content/data/data_transformation_don_hang_clean.csv")
df_clean = df_clean.set_index("OrderID")
df_clean.head()
```

# Select, Rename and Remove Columns

- Chỉ giữ lại các cột cần thiệt và đổi tên cho rõ nghĩa hơn trong bảng phân tích.

```python
df_analysis = df_clean[['City', "Quantity"]].copy()
df_analysis = df_analysis.rename(columns = {"Quantity":"Items"})
df_analysis
```

# Group Data with groupby()

- `groupby()` tạo nhóm, `sum()` tính tổng cho từng nhóm.

```python
city_items = df_analysis.groupby("City")["Items"].sum()
city_items
```

# Create a Derived Column
- Một cột phát sinh tính toán thông tin mới từ các giá trị hiện có

```python
average_items = df_analysis['Items'].mean()
df_analysis["AboveAverage"] = df_analysis['Items'] > average_items
df_analysis
```

# Filter and Sort Rows
- Filter giữa lại các dòng thảo mãn điều kiện.
- Sort thay đổi thứ tự các dòng.

```python
filtered_orders = df_analysis.loc[df_analysis['AboveAverage'],["City", "Items"]]
sorted_orders = filtered_orders.sort_values("Items", ascending=False)
sorted_orders
```

# Summarize Data with agg()
- `agg()` tính nhiều chỉ số tổng hợp cùng lúc cho mỗi nhóm.

```python
df_analysis
```

```python
df_analysis.groupby("City").agg(
    OrderCount = ("Items", "size"),
    TotalItems = ("Items", "sum"),
    AverageItems = ("Items", "mean")
)
```
