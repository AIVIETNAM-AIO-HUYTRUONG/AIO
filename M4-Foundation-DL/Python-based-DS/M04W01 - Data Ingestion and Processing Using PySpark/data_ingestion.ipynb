| **Hồ dữ liệu** | **Data Lake** | Kho lưu trữ chứa lượng lớn dữ liệu thô (Raw Data) ở mọi định dạng (Structured, Semi-structured, Unstructured). |

- Trong các hệ thống **Machine Learning** và **AI** hiện đại, chất lượng của **model** phụ thuộc mạnh vào chất lượng của dữ liệu được sử dụng để huấn luyện và đánh giá model. Một model dù có **architecture** tốt đến đâu cũng khó tạo ra kết quả đáng tin cậy nếu dữ liệu huấn luyện (**training data**) nghèo nàn, thiếu nhất quán, chứa nhiều lỗi hoặc không phản ánh đúng dữ liệu thực tế.

- Vì vậy, trước khi tìm hiểu cách xây dựng và huấn luyện **model**, chúng ta cần hiểu dữ liệu được lấy từ đâu, được thu thập như thế nào và đi qua những bước nào trước khi trở thành dữ liệu có thể sử dụng cho **Machine Learning** và **analytics**.

- Trên thực tế, dữ liệu hiếm khi nằm sẵn ở dạng **ready-to-use**. Dữ liệu thường phát sinh từ nhiều **data sources** khác nhau, chẳng hạn như:

  * Database của doanh nghiệp.
  * Web services và APIs.
  * Files như CSV, JSON, Excel hoặc log files.
  * Thiết bị cảm biến và hệ thống **IoT (Internet of Things)**.
  * Mạng xã hội (**Social Networks**).
  * Các hệ thống và ứng dụng nội bộ.

- Mỗi nguồn dữ liệu có thể sử dụng một định dạng, cấu trúc và cách truy cập khác nhau. Vì vậy, để biến những mảnh dữ liệu rời rạc thành một nguồn dữ liệu thống nhất phục vụ cho **analytics**, **Machine Learning** và các ứng dụng AI, dữ liệu thường phải đi qua một **data pipeline**.

- Ở mức cơ bản, pipeline có thể được hình dung qua ba bước chính:

  1. **Kết nối tới các data sources**
    Xác định và thiết lập kết nối tới các nguồn dữ liệu như database, APIs, files hoặc các hệ thống khác.

  2. **Thu thập và nạp dữ liệu — Data Ingestion**
    Dữ liệu được lấy từ các nguồn và đưa vào hệ thống lưu trữ tập trung (**centralized storage**).

  3. **Lưu trữ dữ liệu**
    Dữ liệu được lưu trong một hệ thống phù hợp để có thể tiếp tục được xử lý, phân tích hoặc sử dụng cho Machine Learning.

- Trong chương này, hệ thống lưu trữ được sử dụng là **Data Lake**. Data Lake cho phép lưu trữ một lượng lớn dữ liệu với nhiều loại định dạng khác nhau, đặc biệt phù hợp với dữ liệu chưa được xử lý hoàn toàn (**raw data**).

- Từ đây, chúng ta có thể hình dung luồng dữ liệu cơ bản như sau:

  **Data Sources → Data Ingestion → Data Lake → Data Processing → Analytics / Machine Learning**

- Đây chính là nền tảng để chúng ta hiểu cách dữ liệu di chuyển trong một hệ thống **Data Pipeline** trước khi được sử dụng để xây dựng các hệ thống Machine Learning và AI.

- Tiếp theo, phần **Data Ingestion** trình bày quá trình **ingest** data từ nguồn và hệ thống, đồng thời giới thiệu ba phương pháp **data ingestion** phổ biến:
  - **Batch ingestion**: Thu thập data theo từng đợt.
  - **Micro-batching**: Thu thập data theo các khoảng thời gian ngắn và batch nhỏ.
  - **Streaming**: Thu thập và xử lý data liên tục khi data phát sinh.

- Sau đó, phần Data Lake giới thiệu về **hệ thống lưu trữ tập trung (centralized storage)**, nơi data có thể được lưu trữ với nhiều định dạng khác nhau, đặc biệt là **raw data**, cùng với một số **platforms** phổ biến được sử dụng để xây dựng **Data Lake**.

- Cuối cùng, phần Hands-on sẽ kết hợp toàn bộ những kiến thức trên để xây dựng một **data pipeline** hoàn chỉnh. **Pipeline** này được tự động hóa để fetch (thu thập) data hằng ngày từ ba **real-world data sources (nguồn data thực tế)**:

  - **arXiv** — data về các bài nghiên cứu khoa học.

- Qua đó, người học có thể theo dõi toàn bộ hành trình của data:

  **Data Sources → Data Ingestion → Data Lake → Automated Data Pipeline**

# **2. Data Source**


| Thuật ngữ tiếng Việt | Thuật ngữ tiếng Anh | Ý nghĩa / Giải thích trong Hệ thống Data |
| :--- | :--- | :--- |
| **Nguồn dữ liệu** | **Data source** | Nơi xuất phát hoặc lưu trữ dữ liệu đầu vào (Database, API, File, Sensor, Logs). |
| **Tạo dữ liệu** | **Data generation** | Quá trình dữ liệu sinh ra tự động từ hành vi người dùng, giao dịch hoặc thiết bị IoT. |
| **Phương thức cung cấp** | **Delivery method / Ingestion mode** | Cách thức dữ liệu được truyền đi (ví dụ: Push/Pull qua API, Batch Export, Event Streaming). |
| **Mức độ cấu trúc** | **Structure level** | Phân loại dữ liệu thành Cấu trúc (Structured), Bán cấu trúc (Semi-structured), và Phi cấu trúc (Unstructured). |
| **Khai thác dữ liệu** | **Data exploitation / Utilization** | Việc sử dụng dữ liệu đã xử lý cho các bài toán Business Intelligence (BI), Dashboard, hoặc Machine Learning. |

- **Data Source** là nguồn cung cấp data cho hệ thống. Một **data source** có thể **store**, **manage**, hoặc **tạo ra (generate)** dữ liệu, đồng thời cung cấp cơ chế data được truy cập hoặc **truyền (transfer/stream)** tới các hệ thống khác nhằm phục vụ **data ingestion** - **processing** - **analytics** và các hoạt động **sử dụng data ở các hệ thống phía sau (downstream use)**.

- **Data Source** có thể phân loại theo ba khía cạnh khác nhau:
  1. **Phân loại theo nguồn gốc - Origin**: nhằm xác định data phát sinh từ đâu, chẳng hạn như **database, API, files, IoT devices, social networks**.
  2. **Phân loại theo cung cấp data - Delivery Method**: nhằm xác định data được đưa vào hệ thống như nào, chẳng hạn như **batch, micro-batching, streaming**.
  3. **Phân loại theo structure level**: nhằm xác định data được tổ chức chặt chẽ đến đâu. Đây là yếu tố ảnh hưởng đến các data được lưu trữ, xử lý và những **tools** cũng như **methods** phù hợp cho **analytics**.

- Do đó, một **Data Source** cụ thể không chỉ thuộc về một nhóm duy nhất. Nó có thể đồng thời được mô tả nhiều loại khác nhau.

- VD: một **YouTube API** được mô tả như sau:
  - **Origin**: Web Service / API
  - **Delivery Method**: Batch
  - **Structure Level**: Semi-structured

- Hiểu cách phân loại này giúp chúng ta hiểu **data source** không chỉ dựa trên việc dữ liệu đến từ đâu, mà còn dựa trên cách dữ liệu được cung cấp và cách dữ liệu được tổ chức.

## **2.1. Phân loại theo nguồn gốc**



### **2.1.1. Database**

- **Databases (cơ sở dữ liệu)** là một trong những **traditional and popular data sources (nguồn dữ liệu truyền thống và phổ biến)**. Chúng được sử dụng để lưu trữ và quản lý dữ liệu có cấu trúc hoặc có schema tương đối rõ ràng, phổ biến trong các hệ **RDBMS (Relational Database Management System)** và **NoSQL databases**.

- Trong doanh nghiệp, các hệ thống **OLTP (Online Transaction Processing)** như **CRM**, **ERP** và **HRM** liên tục ghi nhận các **daily operations (hoạt động hằng ngày)** như tạo đơn hàng, cập nhật thông tin khách hàng, quản lý nhân viên hoặc xử lý giao dịch.

- Các database phục vụ OLTP thường được **optimize (tối ưu)** cho việc xử lý **transaction** nhanh, đảm bảo **data consistency (tính nhất quán dữ liệu)** và **data integrity (tính toàn vẹn dữ liệu)**. Dữ liệu trong các hệ thống này cũng thường được **normalize (chuẩn hóa)** để giảm dư thừa.

- Do đó, việc thực hiện các **analytical queries (truy vấn phân tích)** phức tạp trực tiếp trên hệ thống OLTP có thể tiêu tốn tài nguyên và ảnh hưởng đến **performance (hiệu năng)** của hệ thống giao dịch.

- Vì vậy, trong nhiều kiến trúc dữ liệu, data từ các operational databases được **extract (trích xuất)** thông qua các cơ chế như **ETL**, **ELT**, **CDC (Change Data Capture)** hoặc replication, sau đó được đưa vào **Data Warehouse**, **Data Lake** hoặc **Lakehouse** để phục vụ **analytics**, **reporting** và **Machine Learning**.



### **2.1.2. APIs**

- **APIs (Application Programming Interfaces)** đóng vai trò như một **interface** cho phép các **software applications** giao tiếp và trao đổi data với nhau.

- Thông qua API, một hệ thống có thể **fetch data** theo yêu cầu hoặc định kỳ. Một số API cũng cung cấp cơ chế cập nhật data gần thời gian thực.

- API đặc biệt phổ biến khi cần ingest data từ các **web services, cloud platforms, third-party applications** như:
  - Social network APIs.
  - Payment gateway APIs.
  - Weather APIs.
  - Mapping APIs.
  - External analytics services.

- VD: Một **data pipeline** có thể gọi **YouTube API** mỗi ngày để lấy thông tin về videos, channels,.... rồi lưu trữ vào **Data Lake**.

### **2.1.3. Flat Files**

- **Flat files** là các files đơn giản được sử dụng để lưu trữ hoặc trao đổi dữ liệu mà không yêu cầu một **database management system**.

- Một số định dạng phổ biến gồm:

  - CSV (Comma-Separated Values).
  - TSV (Tab-Separated Values).
  - Excel spreadsheets.
  - JSON (JavaScript Object Notation).
  - XML (Extensible Markup Language).

- **CSV** và **TSV** thường chứa **structured data**, trong khi JSON và XML thường được sử dụng cho **semi-structured data**. Riêng TXT là định dạng rất tổng quát và có thể chứa **structured**, **semi-structured** hoặc **unstructured data** tùy vào nội dung.

- **Flat files** thường được sử dụng để giữa các hệ thống hoặc làm bước trung gian trong các **data pipeline**.

- Ưu điểm của **flat files** là đơn giản và dễ trao đổi. Tuy nhiên, khi file size tăng lớn, việc quản lý, cập nhật, query và phân tích trực tiếp trên files có thể trở nên khó khăn.

### **2.1.4. Streaming Data**

- **Streaming data (dữ liệu luồng)** là dữ liệu được tạo ra liên tục theo thời gian, thường được biểu diễn dưới dạng các **events (sự kiện)**.

- Ví dụ:

    ```text
    User clicks
        ↓
    Event generated
        ↓
    Streaming platform
        ↓
    Data consumer
        ↓
    Processing / Analytics
    ```

- Các hệ thống **streaming** thường sử dụng những công nghệ như **Apache Kafka, RabbitMQ** hoặc các hệ thống hỗ trợ giao thức như **MQTT**, tùy vào yêu cầu kiến trúc.

- Các **data sources** có thể tạo ra **streaming data** gồm:

  - IoT sensors.
  - Transactional systems.
  - Application logs.
  - Online applications.
  - User activity.

- **Streaming data** đặc biệt quan trọng đối với các ứng dụng cần phản hồi **real-time** hoặc **near real-time**, chẳng hạn như **cybersecurity monitoring, fraud detection, system monitoring, live event tracking**.

### **2.1.5. Cloud Services**

- Các **cloud services** ngày càng trở thành nguồn dữ liệu phổ biến trong các hệ thống hiện đại.

- Có thể chia chúng thành nhiều nhóm khác nhau, chẳng hạn:

  - **Cloud object storage:**

  * Amazon S3.
  * Google Cloud Storage.
  * Azure Blob Storage.

  - **Cloud data warehouses:**

  * Google BigQuery.
  * Amazon Redshift.
  * Snowflake.

- Các dịch vụ này cung cấp khả năng **scalability (khả năng mở rộng)**, **high availability (độ sẵn sàng cao)** và khả năng tích hợp với nhiều **data analytics pipelines**.

- Trong một data pipeline, cloud services có thể vừa đóng vai trò là **data source**, vừa là nơi lưu trữ dữ liệu sau khi ingestion.



### **2.1.6. Other Data Sources**

- Ngoài các nguồn dữ liệu có cấu trúc rõ ràng như databases và APIs, data pipeline cũng có thể thu thập dữ liệu từ nhiều nguồn bên ngoài như:
    
  * RSS feeds.
  * Websites.
  * Forums.
  * Social networks.
  * Public datasets.

- Dữ liệu từ các nguồn này thường được **collect (thu thập)** thông qua APIs, feeds hoặc **web scraping (thu thập dữ liệu từ website)**.

- Tùy vào source, dữ liệu có thể ở dạng **structured**, **semi-structured** hoặc **unstructured**.

- Ví dụ, HTML của một website có thể chứa nhiều thông tin bán cấu trúc, trong khi nội dung bài viết hoặc hình ảnh có thể mang tính unstructured cao hơn.

### **2.1.7. Application and System Logs**

- **Logs (nhật ký hệ thống và ứng dụng)** là các records ghi lại những **events** xảy ra trong quá trình **system operation (vận hành hệ thống)**.

- Ví dụ:

    ```text
    User login
    API request
    Database query
    Application error
    Server restart
    File access
    ```

- Logs là một **data source** quan trọng cho:

  * **System monitoring**.
  * **Security monitoring**.
  * **Debugging**.
  * **Performance analysis**.
  * **User behavior analytics**.

- Logs có thể được lưu thành files, database records hoặc được **stream** trực tiếp vào một **logging/streaming system**.

- Điều này cho thấy một data source có thể đồng thời được phân loại theo nhiều dimensions. Ví dụ:

    ```text
    Application Logs

    Origin:
    → Application / System

    Structure:
    → Structured / Semi-structured

    Delivery:
    → Batch hoặc Streaming
    ```

### **2.1.8. IoT and Sensor Data**

- **IoT (Internet of Things)** và **sensor data** là dữ liệu được tạo ra từ các **sensors**, **IoT devices**, GPS, cameras và các thiết bị vật lý khác.

- Các thiết bị này có thể liên tục generate dữ liệu về:

  * Temperature.
  * Humidity.
  * Location.
  * Motion.
  * Device status.
  * Energy consumption.
  * Images hoặc video.

- Ví dụ:

    ```text
    Temperature Sensor
          │
          ▼
    Generate Event
          │
          ▼
    IoT Gateway
          │
          ▼
    Streaming Platform
          │
          ▼
    Data Lake
          │
          ▼
    Analytics / ML
    ```

- Tùy vào yêu cầu của hệ thống, IoT data có thể được truyền theo **real-time**, **near real-time** hoặc được thu thập theo **batch**.

- IoT data được sử dụng rộng rãi trong **monitoring systems**, **smart manufacturing**, **smart cities**, logistics và các hệ thống tự động hóa.

## **2.2. Phân loại Data theo Structure Level**

- Nếu tiêu chí **origin (nguồn gốc)** trả lời cho câu hỏi **data đến từ đâu**, thì tiêu chí **structure (cấu trúc)** trả lời cho câu hỏi **data được organize & represent (tổ chức và biểu diễn) như thế nào**.

- Dựa trên **structure level (mức độ cấu trúc)**, dữ liệu được cung cấp bởi các **data sources** thường được phân thành ba nhóm chính:

### **2.2.1. Structured Data**

- **Structured data (dữ liệu có cấu trúc)** là dữ liệu được tổ chức theo một **predefined schema (lược đồ xác định trước)**, trong đó cấu trúc của dữ liệu được xác định tương đối rõ ràng. Dạng biểu diễn phổ biến nhất là **tabular data**, gồm các **rows & columns (hàng và cột)**.

- Nhờ có cấu trúc và schema rõ ràng, structured data có thể được **store, query & analyze (lưu trữ, truy vấn và phân tích)** hiệu quả bằng các hệ **DBMS/RDBMS**.

- Các nguồn phổ biến gồm:

  * **Relational databases** như MySQL, PostgreSQL và SQL Server.
  * **Structured CSV files**.
  * **Excel spreadsheets**.
  * Các bảng dữ liệu trong **Data Warehouse**.

- Structured data thường được sử dụng để lưu trữ dữ liệu **transactional**, thông tin **customers**, **finance**, **sales** và các dữ liệu phục vụ **daily operations** của doanh nghiệp.

- Ví dụ:

    ```text
    Customer Table

    +----+----------+-----+------------+
    | ID | Name     | Age | City       |
    +----+----------+-----+------------+
    | 01 | An       | 25  | Da Nang    |
    | 02 | Minh     | 30  | Hanoi      |
    | 03 | Huy      | 27  | Hue        |
    +----+----------+-----+------------+
    ```

- Ở đây, mỗi column có một ý nghĩa và data type tương đối rõ ràng.

### **2.2.2 Unstructured Data**

- **Unstructured data (dữ liệu phi cấu trúc)** là dữ liệu không tuân theo một **fixed schema (lược đồ cố định)** và không có cấu trúc dạng bảng rõ ràng để có thể trực tiếp biểu diễn thành các **rows & columns**.

- Các ví dụ phổ biến gồm:

  * **Raw text**.
  * **Emails**.
  * **Social media posts**.
  * **Images**.
  * **Audio**.
  * **Videos**.
  * Documents.

- Do không có cấu trúc cố định, việc **store, search & analyze** unstructured data thường phức tạp hơn structured data và có thể yêu cầu các kỹ thuật xử lý chuyên biệt.

- Unstructured data thường được lưu trong **Data Lake** hoặc **Object Storage**, sau đó được xử lý bằng các kỹ thuật như:

  * **NLP (Natural Language Processing)** — xử lý text và ngôn ngữ tự nhiên.
  * **Computer Vision** — xử lý images và videos.
  * **Machine Learning** — trích xuất patterns hoặc information từ dữ liệu.

- Ví dụ:

  ```text
  Image
     │
     └── pixels
          │
          ▼
  Computer Vision
          │
          ▼
  Features / Information
  ```

- Điểm quan trọng là unstructured không có nghĩa là **“không có bất kỳ structure nào”**. Nó chủ yếu có nghĩa là dữ liệu **không được tổ chức theo một schema cố định, rõ ràng ở cấp độ dữ liệu** như relational table.


### **2.2.3. Semi-Structured Data**

- **Semi-structured data (dữ liệu bán cấu trúc)** nằm giữa structured và unstructured data.

- Loại dữ liệu này không yêu cầu một **fixed schema** chặt chẽ như relational tables, nhưng vẫn chứa các thành phần giúp mô tả và xác định mối quan hệ giữa các dữ liệu, chẳng hạn như **keys**, **values**, **nested objects**, **arrays** hoặc **tags**.

- Các formats phổ biến gồm:

  * **JSON**.
  * **XML**.
  * **HTML**.
  * **YAML**.

- Ví dụ JSON:

  ```json
  {
    "id": 101,
    "name": "Huy",
    "skills": [
      "Python",
      "Machine Learning"
    ]
  }
  ```

- Ta không có một bảng cố định với rows & columns như relational database, nhưng vẫn có các keys như `id`, `name` và `skills` giúp chúng ta hiểu cách dữ liệu được tổ chức.

- Nhờ cấu trúc linh hoạt, semi-structured data thường được sử dụng để **exchange data (trao đổi dữ liệu)** giữa các systems, đặc biệt thông qua APIs và web services.

- Dữ liệu semi-structured cũng có thể được **parsed (phân tích cú pháp)** và **transformed (chuyển đổi)** thành structured data trước khi **load** vào database hoặc **Data Warehouse**.

##**3. Data Ingestion**

### **3.1 Giới thiệu về Data Ingestion**



| Thuật ngữ tiếng Việt | Thuật ngữ tiếng Anh | Ý nghĩa / Giải thích trong Hệ thống Data |
| :--- | :--- | :--- |
| **Nạp dữ liệu** | **Data ingestion** | Quá trình di chuyển dữ liệu từ các nguồn phát sinh về kho lưu trữ tập trung. |
| **Lưu trữ trung tâm** | **Centralized storage** | Nơi tập trung toàn bộ dữ liệu của doanh nghiệp để dễ quản lý, bảo mật và phân tích. |
| **Ứng dụng SaaS** | **SaaS applications** | Các ứng dụng phần mềm chạy trên mây (như Salesforce, Hubspot, Google Analytics) phát sinh dữ liệu cần thu thập. |
| **Luồng thời gian thực**| **Real-time data stream** | Dữ liệu được liên tục tạo ra và truyền đi ngay lập tức (như dòng giao dịch ngân hàng, định vị GPS). |
| **Kho dữ liệu** | **Data Warehouse** | Hệ thống lưu trữ dữ liệu có cấu trúc, đã tối ưu hóa cho truy vấn báo cáo và Business Intelligence (BI). |
| **Kiến trúc Lakehouse** | **Data Lakehouse** | Mô hình lưu trữ hiện đại kết hợp tính linh hoạt của Data Lake và khả năng quản lý ACID/Schema của Data Warehouse. |

- **Data Ingestion** là quá trình **thu thập và đưa data từ các data sources vào hệ thống đích để lưu trữ và xử lý**.

- Các **data sources** có thể rất đa dạng, chẳng hạn như **database, SaaS applications, data files, APIs, IoT devices, real-time data streams**.

- Sau khi được **ingest**, data có thể được đưa trực tiếp vào các hệ thống **storage** như **Data Lake, Data Warehouse** hoặc **Lakehouse**, hoặc được chuyển tiếp đến các **Data processing platforms** để tiếp tục xử lý và biến đổi.

- Điểm quan trọng là **Data Ingestion trả lời câu hỏi: "Làm thế nào để đưa data từ source vào hệ thống?"**

- Trong khi đó:
  - **Data Source**: Data đến từ đâu?
  - **Data Ingestion**: Data được đưa vào hệ thống bằng cách nào?
  - **Data Storage**: Data được lưu ở đâu?
  - **Data Processing**: Data được xử lý như thế nào?

- Tùy vào đặc điểm của data và kiến trúc hệ thống, data sau khi được thu thập có thể đưa vào **Data Lake, Data Warehouse, Lakehouse** hoặc **Data processing platforms** khác.

### **Lưu ý: Data Ingestion và ETL**

- Khác với quy trình **ETL** phải xử lý và biến đổi data trước khi lưu, **data ingestion** đẩy **raw data** vào hệ thống đích.

- Nhờ bỏ qua bước biến đổi trung gian, việc đưa data vào hệ thống diễn ra nhanh và linh hoạt hơn, đổi lại data cần được làm sạch và chuẩn hóa ở các bước sau.

### **3.2. Các phương pháp Data Ingestion**

| Thuật ngữ tiếng Việt | Thuật ngữ tiếng Anh | Ý nghĩa / Giải thích trong Hệ thống Data |
| :--- | :--- | :--- |
| **Độ trễ** | **Latency** | Khoảng thời gian trễ từ lúc dữ liệu phát sinh ở nguồn cho đến khi sẵn sàng để sử dụng/truy vấn. |
| **Bắt dữ liệu thay đổi**| **Change Data Capture (CDC)** | Kỹ thuật theo dõi và ghi lại các hành động INSERT, UPDATE, DELETE trong Database để đồng bộ sang hệ thống khác. |
| **Kích hoạt theo sự kiện**| **Trigger-based / Event-driven** | Tiến trình chạy tự động dựa trên một sự kiện cụ thể xảy ra (ví dụ: file mới vừa được tải lên S3). |
| **Đường ống độ trễ thấp**| **Low-latency pipeline** | Quy trình xử lý và di chuyển dữ liệu với khoảng thời gian chờ cực ngắn (milisecond đến giây). |
| **Khối lượng công việc**| **Workload** | Mức độ tải hoặc số lượng tác vụ/truy vấn mà hệ thống/Database đang phải gánh vác xử lý. |

- Tùy theo yêu cầu về **độ trễ (latency)**, tần suất cập nhật và đặc điểm của **workload**. **Data Ingestion** thường được triển khai theo ba phương pháp chính:
  - **Batch ingestion**
  - **Micro-batching**
  - **Streaming ingestion**

#### **3.2.1. Batch Ingestion**

- **Batch ingestion (nạp data theo lô)** là phương pháp thu thập và di chuyển dữ liệu thành từng **batch (lô)** tại các khoảng thời gian được định trước, chẳng hạn như mỗi vài phút, mỗi giờ, mỗi ngày hoặc mỗi tuần.

- Thay vì xử lý từng **data record** ngay khi nó phát sinh, hệ thống sẽ tích lũy một lượng data nhất định rồi xử lý toàn bộ trong một lần. Việc chạy **batch** thường được thực hiện theo **schedule (lịch trình)** hoặc được kích hoạt bởi một **event** hay một điều kiện cụ thể.

- VD: Doanh nghiệp có thể thiết lập **pipeline** để mỗi ngày lúc **00:00**, **ingest** toàn bộ data giao dịch của ngày hôm trước từ hệ thống CRM hoặc ERP vào **Data Lake**.

- **Batch ingestion** phù hợp với những **use cases** không yêu cầu data được cập nhật ngay lập tức, chẳng hạn như:
  - **Daily reporting**
  - **Data warehouse loading**
  - **historical analytics**
  - hoặc các pipeline xử lý dữ liệu với khối lượng lớn.

#### **3.2.2. Micro-batching**

- **Micro-batching (nạp data theo các lô nhỏ)** là phương pháp giữa **batch ingestion** và **streaming ingestion**. Thay vì chờ một khoảng thời gian dài để tích lũy một **batch lớn**, hệ thống sẽ liên tục gom data thành các **batch nhỏ** và xử lý chúng với khoảng thời gian ngắn hơn.

- Khoảng thời gian giữa các batches có thể từ vài giây đến vài phút, tùy thuộc vào yêu cầu của độ trễ và kiến trúc hệ thống.

- Cách tiếp cận này giúp giảm thời gian chờ so với xử lý theo lô truyền thống, đồng thời vẫn cho phép hệ thống xử lý data theo từng nhóm nhỏ thay vì phải xử lý theo từng **record riêng lẻ**

#### **3.2.3. Streaming Ingestion**

- **Streaming ingestion (nạp data liên tục)** là phương pháp đưa data vào hệ thống gần như ngay lập tức khi các **events** phát sinh, thay vì chờ data được tích thành một **batch**.

- **Streaming ingestion** phù hợp với các **trường hợp yêu cầu phản ứng nhanh**, nơi hệ thống cần phát hiện và xử lý thông tin mới với **độ trễ thấp**, chẳng hạn như:
  - **Fraud Detection**
  - **Application Monitoring**
  - **IoT Telemetry**
  - **Real-time Recommendations**
  - hoặc các hệ thống giao dịch

- Một kỹ thuật thường được sử dụng trong **streaming architecture** là **Change Data Capture (CDP)**. CDC theo dõi các thay đổi xảy ra trong database, chẳng hạn như **INSERT, UPDATE, DELETE** rồi phát ra các thay đổi đó thành các event để **downstream systems** có thể tiếp tục xử lý.

#### **So sánh ba phương pháp**

Có thể hình dung ba phương pháp dựa trên câu hỏi:

> **“Chúng ta chờ bao lâu trước khi đưa dữ liệu vào pipeline?”**

```text
Batch
│
│  Chờ lâu → gom nhiều data
│
├──────────────────────────────►
│
│
Micro-batching
│
│  Chờ ngắn → gom các batch nhỏ
│
├──────► ├──────► ├──────►
│
│
Streaming
│
│  Gần như xử lý ngay khi event xuất hiện
│
├─► ├─► ├─► ├─► ├─► ├─►
│
└──────────────────────────────────────► Time
```

| Method             | Data được xử lý      | Latency        | Use cases điển hình                                 |
| ------------------ | -------------------- | -------------- | --------------------------------------------------- |
| **Batch**          | Theo batch lớn       | Cao            | Daily reports, ETL, historical analytics            |
| **Micro-batching** | Theo batch nhỏ       | Thấp hơn batch | Fleet tracking, periodic updates                    |
| **Streaming**      | Liên tục theo events | Rất thấp       | Monitoring, fraud detection, IoT, real-time systems |

Như vậy, sự khác biệt cốt lõi giữa ba phương pháp không đơn giản là **“nhanh hay chậm”**, mà nằm ở **cách hệ thống thu thập, gom nhóm và đưa dữ liệu vào pipeline theo thời gian**

## **4. Data Lake**

- **Data Lake** là một **hệ thống lưu trữ tập trung** được thiết kế để lưu trữ khối lượng data lớn với nhiều loại và nhiều định dạng khác nhau.

-  **Data Lake** được sử dụng như một nơi tiếp nhận và lưu trữ data với mức độ biến đổi ban đầu thấp, trong đó **raw data** là một loại data phổ biến.

- Trong các **modern data system**, **Data Lake** thường được xây dựng trên các nền tảng **Object Storage** như:
  - **Amazon S3**
  - **Azure Blob Storage**
  - **Google Cloud Storage**

  - Các nền tảng này cung cấp khả năng **scalability (khả năng mở rộng linh hoạt), durability (độ bền data)** và **cost-effective storage (lưu trữ với chi phí hiệu quả)**, phù hợp với việc lưu trữ khối lượng data lớn.


- Một đặc điểm quan trọng của **Data Lake** là khả năng tiếp nhận nhiều loại dữ liệu khác nhau. Không giống với các hệ thống yêu cầu một **schema** cố định trước khi ghi data, **Data Lake** thường áp dụng mô hình **schema-on-read**: cấu trúc và cách diễn giải data được xác định khi data được đọc hoặc xử lý. Nhờ đó, **Data Lake** có thể lưu trữ đồng thời:

  - **Structured data**: data dạng bảng từ **database**, **CSV** hoặc các hệ thống nghiệp vụ khác.
  - **Semi-structured data**: JSON, XML, Logs hoặc API response.
  - **Unstructured data**: text documents, images, audio, video,...

### **4.1. Vai trò của Data Lake**

- **Data Lake** có thể đảm nhận nhiều roles trong một **data architecture**:

#### **4.1.1. Centralized Storage**

- **Data Lake** cung cấp một nơi lưu trữ tập trung cho data đến từ nhiều **data sources** khác nhau. Thay vì mỗi hệ thống lưu trữ data hoàn toàn tách biệt, data có thể đưa về một nền tảng lưu trữ chung để phục vụ các **downstream workloads**.

#### **4.1.2. Landing zone**

- Data Lake thường đóng vai trò **landing zone** (khu vực tiếp nhận dữ liệu), nơi dữ liệu mới được đưa vào sau quá trình **data ingestion** trước khi được **processing**, cleaning hoặc transformation.

- Ví dụ:

    ```text
    API ────────┐
    Database ───┤
    Files ──────┼──► Data Ingestion ──► Data Lake
    IoT ────────┘                              │
                                              │
                                          Processing
                                              │
                                              ▼
                                      Analytics / ML / AI
    ```



#### **4.1.3. Historical data storage**

- Data Lake có thể lưu trữ lượng lớn **historical data** (dữ liệu lịch sử), bao gồm cả dữ liệu cũ ít được truy cập thường xuyên. Một phần dữ liệu như vậy có thể được quản lý dưới dạng **cold data** (dữ liệu ít được truy cập) để tối ưu chi phí lưu trữ.


#### **4.1.4. Foundation for Data Science and AI**

- Data Lake có thể cung cấp dữ liệu cho nhiều downstream workloads như **Data Science**, **Machine Learning**, **AI** và **Big Data analytics**. Các datasets được lưu trong Data Lake có thể tiếp tục được processing và transformation để tạo ra dữ liệu phù hợp cho từng workload.


### **4.2. Data Lake và Data Swamp**

- Data Lake có khả năng lưu trữ dữ liệu lớn và đa dạng, nhưng khả năng lưu trữ lớn không đồng nghĩa với khả năng quản lý tốt.

- Nếu dữ liệu liên tục được đưa vào mà không có **Data Governance** (quản trị dữ liệu), **Data Catalog** (danh mục dữ liệu), metadata, naming conventions và các cơ chế quản lý phù hợp, Data Lake có thể dần trở thành một **Data Swamp** (đầm lầy dữ liệu).

  ```text
                DATA LAKE
                    │
          ┌─────────┴─────────┐
          │                   │
    Good Governance      Poor Governance
          │                   │
          ▼                   ▼
    Organized Lake       "Data Swamp"
          │                   │
          │                   ├── Hard to find
          │                   ├── Hard to understand
          │                   ├── Hard to manage
          │                   └── Hard to trust
          ▼
  Easy to discover
  Easy to manage
  Easy to use
  ```

### **4.3. Các nền tảng Object Storage cho Data Lake**

- Trên thực tế, nhiều **Data Lake** hiện đại được triển khai trên **Object Storage infrastructure** (hạ tầng lưu trữ đối tượng) của các **major cloud providers** (nhà cung cấp đám mây lớn). Bên cạnh storage layer, kiến trúc Data Lake thường kết hợp thêm các **governance layers** (lớp quản trị), **metadata catalogs** (catalog metadata) và **Open Table Formats** (định dạng bảng mở) như **Delta Lake**, **Apache Iceberg** và **Apache Hudi** để quản lý dữ liệu hiệu quả hơn.

- Một số **platforms** tiêu biểu gồm:


#### **4.3.1 Amazon S3 (Simple Storage Service)**

* **Amazon S3 (Simple Storage Service):** Dịch vụ **Object Storage** của AWS, thường được sử dụng làm storage layer cho Data Lake nhờ khả năng mở rộng lớn, độ bền dữ liệu cao và mô hình chi phí phù hợp với việc lưu trữ khối lượng dữ liệu lớn. S3 có thể kết hợp với các dịch vụ trong hệ sinh thái AWS như **AWS Lake Formation** để quản lý quyền truy cập và governance, **AWS Glue** để catalog và tích hợp dữ liệu, cùng **Amazon Athena** để query dữ liệu trực tiếp trên S3.


#### **4.3.2. Google Cloud Storage (GCS)**

- **Google Cloud Storage (GCS):** Dịch vụ **Object Storage** của Google Cloud, có thể đóng vai trò **central storage layer** cho Data Lake trên Google Cloud. GCS tích hợp với hệ sinh thái **analytics** và **data processing** của Google Cloud, chẳng hạn như **BigQuery** và **Dataproc**, phù hợp với các **large-scale analytics** và **machine learning workloads**.




#### **4.3.3. Azure Data Lake Storage (ADLS)**

- **Azure Data Lake Storage (ADLS):** Dịch vụ lưu trữ dữ liệu quy mô lớn của Microsoft Azure, được xây dựng trên Azure Blob Storage và hỗ trợ **hierarchical namespace** (không gian tên phân cấp). Tính năng này cho phép dữ liệu được tổ chức theo cấu trúc thư mục và cung cấp các khả năng quản lý phù hợp với các **Big Data analytics workloads**. ADLS có thể tích hợp với các dịch vụ như **Azure Synapse Analytics** và **Azure Databricks**.

#### **4.3.4 MinIO**

- **MinIO:** Một giải pháp **open-source Object Storage** tương thích với **Amazon S3 API**. MinIO có thể được triển khai trên **on-premises infrastructure** (hạ tầng tại chỗ), private cloud hoặc hybrid cloud, cho phép tổ chức xây dựng storage layer có khả năng kiểm soát hạ tầng và dữ liệu cao hơn. Đây cũng là một lựa chọn khi tổ chức muốn giảm sự phụ thuộc vào một cloud provider cụ thể và hạn chế **vendor lock-in** (phụ thuộc vào nhà cung cấp).


- Điểm chung của các nền tảng trên là sử dụng **Object Storage model** để cung cấp một storage layer có khả năng mở rộng lớn và hỗ trợ nhiều loại dữ liệu, từ **structured data**, **semi-structured data** đến **unstructured data**.

  ```text
                      DATA SOURCES
                          │
                          ▼
                    DATA INGESTION
                          │
                          ▼
                ┌─────────────────────┐
                │    OBJECT STORAGE   │
                │                     │
                │  Amazon S3          │
                │  Google Cloud       │
                │  Storage             │
                │  Azure ADLS         │
                │  MinIO              │
                └─────────────────────┘
                          │
                ┌──────────┼──────────┐
                ▼          ▼          ▼
            Catalog    Governance   Table Format
                │          │          │
                └──────────┼──────────┘
                          ▼
                Analytics / ML / AI
  ```

- Vì vậy, trong kiến trúc **Data Lake**, **Object Storage** thường đóng vai trò là **storage foundation** (nền tảng lưu trữ), trong khi các thành phần như **catalog**, **governance**, **processing engines** và **Open Table Formats** bổ sung các khả năng cần thiết để biến storage layer thành một hệ thống Data Lake có thể quản lý và khai thác hiệu quả.
